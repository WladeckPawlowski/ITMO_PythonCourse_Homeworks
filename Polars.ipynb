```python
import polars as pl

# Чтение данных
file_path = 'train (3).csv'
df = pl.read_csv(file_path)

# Вывод основной информации
df_info = {
    "dtypes": df.schema,
    "null_count": df.null_count(),
    "mean": df.select(pl.col(pl.NUMERIC_DTYPES).mean())
}

# Подсчет количества пассажиров каждого класса
pclass_counts = df['Pclass'].value_counts()

# Количество выживших мужчин и женщин
survived_gender_counts = df.group_by("Sex").agg(pl.col("Survived").sum().alias("SurvivedCount"))

# Пассажиры старше 44 лет
passengers_over_44 = df.filter(pl.col("Age") > 44)

# Вывод результатов
df_info, pclass_counts, survived_gender_counts, passengers_over_44.head()
```
