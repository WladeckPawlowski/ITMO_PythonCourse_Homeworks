```python
import requests
import pandas as pd

# Выполнение запроса и получение данных
def fetch_data(url):
    response = requests.get(url)
    if response.status_code == 200:
        return response.json()
    else:
        print(f"Ошибка при запросе данных: {response.status_code}")
        return None

# Получение информации о книгах
books_url = "https://anapioficeandfire.com/api/books"
books_data = fetch_data(books_url)
books_df = pd.DataFrame(books_data)

# Получение информации о домах Вестероса
houses_url = "https://anapioficeandfire.com/api/houses"
houses_data = fetch_data(houses_url)
houses_df = pd.DataFrame(houses_data)

# Получение информации о домах с девизом
houses_with_titles_url = "https://anapioficeandfire.com/api/houses?hasTitles=true"
houses_with_titles_data = fetch_data(houses_with_titles_url)
houses_with_titles_df = pd.DataFrame(houses_with_titles_data)

# Сохранение данных в CSV файлы
books_df.to_csv('books.csv', index=False)
houses_df.to_csv('houses.csv', index=False)
houses_with_titles_df.to_csv('houses_with_titles.csv', index=False)

print("Данные успешно сохранены в файлы books.csv, houses.csv и houses_with_titles.csv")
```
