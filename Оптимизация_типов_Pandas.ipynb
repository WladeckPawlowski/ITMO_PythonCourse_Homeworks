```python
import pandas as pd

# Чтение данных из файла
file_path = 'Housing.csv'
df = pd.read_csv(file_path)

# Вывод информации о типах данных и потреблении памяти до оптимизации
memory_before = df.memory_usage(deep=True)
dtypes_before = df.dtypes

# Определение оптимальных типов данных для каждого столбца
optimized_dtypes = {
    'price': 'float32', # Цены - самые большие числа в датасете и могут теоретически содержать дробную часть --> использован тип данных float32
    'area': 'int16',
    'bedrooms': 'int8', # В датасете, значение представляет собой однозначное (целое) число, поэтому было принято решение об использовании типа int8
    'bathrooms': 'int8', # В датасете, значение представляет собой однозначное (целое) число, поэтому было принято решение об использовании типа int8
    'stories': 'int8', # В датасете, значение представляет собой однозначное (целое) число, поэтому было принято решение об использовании типа int8
    'mainroad': 'category', # Ограниченное (не числовое) количество значений, поэтому использован тип категориальной переменной
    'guestroom': 'category', # Ограниченное (не числовое) количество значений, поэтому использован тип категориальной переменной
    'basement': 'category', # Ограниченное количество значений, поэтому использован тип категориальной переменной
    'hotwaterheating': 'category', # Ограниченное (не числовое) количество значений, поэтому использован тип категориальной переменной
    'airconditioning': 'category', # Ограниченное (не числовое) количество значений, поэтому использован тип категориальной переменной
    'parking': 'int8', # В датасете, значение представляет собой (целое) однозначное число, поэтому было принято решение об использовании типа int8
    'prefarea': 'category', # Ограниченное (не числовое) количество значений, поэтому использован тип категориальной переменной
    'furnishingstatus': 'category' # Ограниченное (не числовое) количество значений, поэтому использован тип категориальной переменной
}

# Изменение типов данных столбцов
df_optimized = df.astype(optimized_dtypes)

# Вывод информации о типах данных и потреблении памяти после оптимизации
memory_after = df_optimized.memory_usage(deep=True)
dtypes_after = df_optimized.dtypes

# Сравнение потребления памяти до и после оптимизации
memory_comparison = pd.DataFrame({
    'Before Optimization': memory_before,
    'After Optimization': memory_after})

# Вывод результатов
print("Memory usage before optimization:")
print(memory_before)
print("\nMemory usage after optimization:")
print(memory_after)
print("\nData types before optimization:")
print(dtypes_before)
print("\nData types after optimization:")
print(dtypes_after)
print("\nMemory usage comparison:")
print(memory_comparison)
```
