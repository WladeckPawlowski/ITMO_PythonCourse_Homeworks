```python
import pandas as pd
import bottleneck as bn

# Чтение данных
file_path = 'train (3).csv'
df = pd.read_csv(file_path)

# Расчет среднего возраста пассажиров и его стандартного отклонения
mean_age = bn.nanmean(df['Age'])
std_age = bn.nanstd(df['Age'])

# Умножение значения столбца Fare на 1.3 и сохранение результатов в столбце Fare_new
df['Fare_new'] = df['Fare'].apply(lambda x: x * 1.3)

# Вывод результатов
mean_age, std_age, df[['Fare', 'Fare_new']].head()
```
